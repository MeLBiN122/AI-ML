```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder,MinMaxScaler,StandardScaler
```

```python
df = pd.read_csv("/content/titanic uncleaned.csv")
```

```python
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed'] = df['Age'].fillna(df['Age'].median())
df['Cabin_mode_imputed'] = df['Cabin'].fillna(df['Cabin'].mode()[0])
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))
```

```python
en = df.copy()
label_encoder = LabelEncoder()
en['Sex_Label'] = label_encoder.fit_transform(en['Sex'])
one_hot_encoded = pd.get_dummies(en['Embarked'],prefix='Embarked')
en = pd.concat([en,one_hot_encoded],axis=1)
pclass_mapping = {1: 'First', 2: 'Second', 3: 'Third'}
en['Pclass_Category'] = en['Pclass'].map(pclass_mapping)
```

```python
numerical_columns = ['Age','Fare','Pclass']
```

```python
df.dropna(subset = numerical_columns,inplace = True)
```

```python
min_max_scalar = MinMaxScaler()
standard_scalar = StandardScaler()
```

```python
df['Age_MinMax'] = min_max_scalar.fit_transform(df[['Age']])
df['Fare_MinMax'] = min_max_scalar.fit_transform(df[['Fare']])
```

```python
df['Age_Standard'] = standard_scalar.fit_transform(df[['Age']])
df['Fare_Standard'] = standard_scalar.fit_transform(df[['Fare']])
```
